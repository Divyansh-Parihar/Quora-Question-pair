```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
Quora_Question = pd.read_csv("train.csv")
```

```python
sample = Quora_Question.sample(30000)
```

```python
sample.isnull().sum()
```

```python
sample.duplicated().sum()
```

```python
quest_df = sample[['question1','question2']]
quest_df.head()
```

```python
from sklearn.feature_extraction.text import CountVectorizer

# merge text
questions = list(quest_df['question1']) + list(quest_df['question2'])

cv = CountVectorizer(max_features=3000)
q1_arr, q2_arr = np.vsplit(cv.fit_transform(questions).toarray(),2)
```

```python
temp_df1 = pd.DataFrame(q1_arr,index=quest_df.index)
temp_df2 = pd.DataFrame(q2_arr,index=quest_df.index)
temp_df = pd.concat([temp_df1,temp_df2],axis=1)
temp_df.shape
```

```python
temp_df['is_duplicate'] = sample['is_duplicate']
```

```python
temp_df
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(temp_df.iloc[:,0:-1].values, temp_df.iloc[:,-1].values, test_size=0.2, random_state=42)
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score


rf = RandomForestClassifier()
rf.fit(X_train,y_train)
model = rf.predict(X_test)

print(accuracy_score(model,y_test))
```
