```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
```

```python
Quora_Question = pd.read_csv('train.csv')
```

```python
Quora_Question.head()
```

```python
Quora_Question.info()
```

```python
Quora_Question.isnull().sum()  #missing values
```

```python
Quora_Question.duplicated().sum()  #duplicates
```

```python
print(Quora_Question['is_duplicate'].value_counts())  #Distribution of similar and non-similar Questions
Quora_Question['is_duplicate'].value_counts().plot(kind='bar')
```

```python
#Repeated Questions

qid = pd.Series(Quora_Question['qid1'].tolist() + Quora_Question['qid2'].tolist())
print('Number of Unique Questioins in the DataSet:', np.unique(qid).shape[0])
x = qid.value_counts()>1
print('Number of Repeated Questions:',x.sum())
```

```python
# Repeated Questions Visualization

fig, ax = plt.subplots(figsize=(25,8))

ax.hist(qid.value_counts().values, bins=200)
ax.set_yscale('log')
ax.set_xlabel("Number of Occurrences of the Questions")
ax.set_ylabel("Count of Questions")

plt.tight_layout()
plt.show()
```
